Import Libraries

```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
```

Load Cleaned Data

```python
df = pd.read_csv("../data/processed/marketing_campaign_clean.csv")

df.head()
```

Customer Age

```python
current_year = 2026

df["Age"] = current_year - df["Year_Birth"]
```

Customer Tenure (Days)

```python
df["Dt_Customer"] = pd.to_datetime(df["Dt_Customer"])

df["Customer_Tenure"] = (
    pd.Timestamp("2026-01-01") - df["Dt_Customer"]
).dt.days
```

Family Size

```python
df["Family_Size"] = df["Kidhome"] + df["Teenhome"] + 1
```

Total Spending

```python
df["Total_Spending"] = (
    df["MntWines"] +
    df["MntFruits"] +
    df["MntMeatProducts"] +
    df["MntFishProducts"] +
    df["MntSweetProducts"] +
    df["MntGoldProds"]
)
```

Total Purchases

```python
df["Total_Purchases"] = (
    df["NumWebPurchases"] +
    df["NumCatalogPurchases"] +
    df["NumStorePurchases"]
)
```

Total Accepted Campaigns

```python
df["Accepted_Campaigns"] = (
    df["AcceptedCmp1"] +
    df["AcceptedCmp2"] +
    df["AcceptedCmp3"] +
    df["AcceptedCmp4"] +
    df["AcceptedCmp5"]
)
```

Total Children

```python
df["Children"] = df["Kidhome"] + df["Teenhome"]
```

Save Processed Dataset

```python
df.to_csv("../data/processed/marketing_campaign_feature_engineered.csv", index=False)
```
